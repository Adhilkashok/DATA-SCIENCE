```python
sns.scatterplot(data=df,x="total_bill",y="tip")
plt.show()        # as total bill increases tip increases
```

```python
 #! hue: Used to represent another categorical variables using diff colours
# hue="sex"
sns.scatterplot(data=df,x="total_bill",y="tip",hue="sex")
plt.show()        
```

```python
# hue="smoker"
sns.scatterplot(data=df,x='total_bill',y="tip",hue="smoker")
plt.show()        
```

```python
#  #! style: Is used to separate categories using different marker/shapes
sns.scatterplot(data=df,x='total_bill',y="tip",hue="sex",style="smoker")
plt.show()   
```

```python
 #! size: Is used to change the size of points/markers based on a column.
sns.scatterplot(data=df,x='total_bill',y="tip",hue="sex",style="smoker",size="size")
plt.show()   
```

```python
sns.lineplot(data=df,x="total_bill",y="tip")
plt.show()
```

```python
df["day"].value_counts()
```

```python
# daywise total bill
sns.barplot(data=df,x="day",y="total_bill",errorbar="sd")
plt.show()
 #! line on the bar -> error/uncertainity interval around the estimate.
 #! errorbar: An error bar is used to show the uncertainty, variation, or error in data values.
```

```python
 #! estimator: Changing aggregation. Estimator tells Seaborn which calculation to perform on the values before creating the plot.
import numpy as np
sns.barplot(data=df,x="day",y="total_bill",estimator=np.sum,errorbar=None)   # sum is used in numpy so import numpy
plt.show()
```

```python
 #! countplot: Counts no.of observations in each category
sns.countplot(data=df,x="day")
plt.show()
```

```python
 #! the no of male nd female customr on each day
sns.countplot(data=df,x="day",hue='sex')
plt.show()
```

```python
 #! Histogram: Distribution of numerical data
#? total_bill distribution
sns.histplot(data=df,x="total_bill",bins=10)
plt.show()
```

```python
 #! KDE: Kernal Density Estimate
 #! represents distribution of numerical data as a smooth curve
sns.kdeplot(data=df,x="total_bill")
plt.show()
```

```python
 #!combining KDE in histogram
sns.histplot(data=df,x="total_bill",bins=10,kde=True)
plt.show()
```

```python
sns.boxplot(data=df,x="total_bill")
plt.show()
```

```python
sns.boxplot(data=df,x="total_bill",y="day")
plt.show()
```

```python
sns.boxplot(data=df,x="total_bill",y="day",hue="sex")
plt.show()
```

```python
 #! violin plot: A violin plot is a statistical graph that shows both the distribution of numerical data and its summary statistics. It’s similar to a box plot, but it also shows the shape/density of the data
sns.violinplot(data=df,x="day",y="total_bill")
plt.show()
```

```python
 #! pairplot: shows rlshnps btw multiple numerical variables
sns.pairplot(df)
plt.show()
```

```python
 #? correlation
#! corr() calculates the correlation coefficient between numerical columns
corr=df.corr(numeric_only=True)
corr
```

```python
 #! heatmap:visualizes those correlations using colors
sns.heatmap(corr)
plt.show()
```

```python
 #! annot=True: Used To get the corresponding corr values
sns.heatmap(corr,annot=True)
plt.show()
```
