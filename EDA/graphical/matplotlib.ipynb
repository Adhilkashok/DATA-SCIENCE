```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,15,20,25,30]
plt.plot(x,y)   # -> plot for creating graph
plt.show()   # -> Displays the graph
```

## 1) Line Chart  => Mainly used to show trends,growth,decline,changes over time

```python
days=[1,2,3,4,5]
temperature=[30,32,38,29,31]
plt.plot(days,temperature)

#! Adding Title to chart
plt.title("Daily Temperature")

#! Lbelling X-axis
plt.xlabel("Day")

#! Labelling Y-axis
plt.ylabel("Temperature")

#! Adding Grid Lines => To read data points easily
plt.grid()

plt.show()
```

```python
 #! Changing Line style
  #* => linestyle="-,--,:,-."

days=[1,2,3,4,5]
temperature=[30,32,38,29,31]
plt.plot(days,temperature,linestyle="--")

plt.title("Daily Temperature")

plt.xlabel("Day")

plt.ylabel("Temperature")

plt.grid()

plt.show()
```

```python
 #! Changing line width
   #* linewidth=number

days=[1,2,3,4,5]
temperature=[30,32,38,29,31]
plt.plot(days,temperature,linestyle="--",linewidth=4)

plt.title("Daily Temperature")

plt.xlabel("Day")

plt.ylabel("Temperature")

plt.grid()

plt.show()
```

```python
 #! Mark Data points
   #* ● Circle	marker='o'
   #* ■ Square	marker='s'
   #* ▲ Triangle	marker='^'
   #* ▼ Triangle	marker='v'
   #* ★ Star	marker='*'
   #* + Plus	marker='+'
   #* × X	marker='x'
   #* ◆ Diamond	marker='D'

days=[1,2,3,4,5]
temperature=[30,32,38,29,31]
plt.plot(days,temperature,linestyle="--",linewidth=2,marker="^",color="red")
                                                                #* color=""  -> changing colour of line

plt.title("Daily Temperature")

plt.xlabel("Day")

plt.ylabel("Temperature")

plt.grid()

plt.show()
```

```python
 #! figure() -> Create new figure
days=[1,2,3,4,5]
temperature=[30,32,38,29,31]
plt.figure(figsize=(4,3))  #* width,height
plt.plot(days,temperature)
plt.show()
```

```python
 #! savefig() -> is used to save a graph/figure as an image file.

days=[1,2,3,4,5]
temperature=[30,32,38,29,31]
plt.plot(days,temperature,linestyle="--",linewidth=2,marker="D",color="gold")

plt.title("Daily Temperature")

plt.xlabel("Day")

plt.ylabel("Temperature")

plt.grid()
plt.savefig("temperature.png")

plt.show()
```

```python
month=["jan","feb","mar","apr","may"]
sales=[10000,12000,13000,14000,16000]
plt.plot(month,sales,linestyle="--",linewidth=3,marker="o",color="green")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.savefig("monthlysales.png")
plt.show()
```

```python
 #! Multiple Line Chart
month=["jan","feb","mar","apr","may"]
product_a=[100,125,130,142,150]
product_b=[90,103,120,140,147]
plt.plot(month,product_a)
plt.plot(month,product_b)
plt.grid()
plt.show()
```

```python
 #! Labels and Legend -> labels tell you what the axes represent, while a legend tells you which line/data belongs to which category.
month=["jan","feb","mar","apr","may"]
product_a=[100,125,130,142,150]
product_b=[90,103,120,140,147]
plt.plot(month,product_a,label="Product a")
plt.plot(month,product_b,label="Product b")
plt.legend()
plt.grid()
plt.show()
```

## Bar Charts -> Used to compare values btw categories

```python
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.bar(products,sales)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
 #! Horizontal Bar Chart  => use barh()
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.barh(products,sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
 #! Change Colour
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.bar(products,sales,color="Red")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
 #! Change Bar Width
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.bar(products,sales,width=0.2)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
 #! Add Border => edgecolor
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.bar(products,sales,edgecolor="red")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
 #! Combining both
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.bar(products,sales,color="red",edgecolor="yellow")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
 #! transparency => alpha
#? ranges from 0-1
products=["A","B","C","D"]
sales=[100,105,102,120]
plt.bar(products,sales,alpha=0.3)
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

## scatter plot: A scatter plot is used to show the relationship between two numerical variables.

```python
hour=[1,2,3,4,5]
marks=[38,42,49,55,60]
plt.scatter(hour,marks)
plt.show()

#* Positive relation: if hour increases marks also increases
#* Negative relation: as pressure increases volume decreases
#* No relation: When there is no clear relationship between X and Y
```

## Histogram: A histogram is used to show the distribution of numerical data by dividing the data into intervals called bins.

```python
ages=[20,21,22,27,24,25,22,24,21,20,19]
plt.hist(ages,bins=6)
plt.show()

#! bins: histogram divides values into intervals called bins
```

```python
 #! 📊 Difference Between Bar Chart and Histogram
# Bar chart is used to compare different categories
# Histogram is used to show the distribution of numerical data
# Bar chart can contain gaps between bars
# Histogram bars usually touch each other
# Bar chart uses categories on the X-axis
# Histogram uses numerical intervals (bins) on the X-axis
# Bar chart → Comparison
# Histogram → Distribution
```

## piechart: It represents parts of a whole

```python
lang=["python","R","C","Java"]
hours=[40,20,10,30]
plt.pie(hours,labels=lang)
plt.show()
```

```python
 #! autopct → displays percentages
lang=["python","R","C","Java"]
hours=[40,20,10,30]
plt.pie(hours,labels=lang,autopct="%1.1f%%")         # %1.1f → shows 1 decimal place
                                                     # %% → displays the % symbol
plt.show()
```

```python
 #! explode: used to separate/highlight a particular slice of the pie chart.
explode=[0,0,0.1,0]    # here 0.1 will be highlighted. Those which need not to be highlighted will be marked as 0
lang=["python","R","C","Java"]
hours=[40,20,10,30]
plt.pie(hours,labels=lang,autopct="%1.1f%%",explode=explode)
#* or
# plt.pie(hours,labels=lang,autopct="%1.1f%%",explode=[0,0,0.1,0])
plt.show()
```

## boxplot: Summarizes the distribution of numerical data
## It is useful for identifying outliers
## It uses
          Q1
          Q2 - Median
          Q3
          Maximum
          Minimum

```python
ages=[20,2,28,24,25,200,289,26]
plt.boxplot(ages)
plt.show()

#   289  ●  ← Outlier
#        |
#   200  ●  ← Outlier
#        |
#   28 ──┐
#        │
#   26 ──┤
#   25 ──┤ ← Median ≈ 25.5
#   24 ──┤
#   20 ──┘
#        |
#    2 ──┘
```

## subplot: Subplot is used to display multiple plots in one figure.
##          It divides one figure into rows and columns.

```python
 #* syntax: plt.subplot(rows, columns, position)
plt.subplot(2,2,1)
plt.boxplot(ages)

plt.subplot(2,2,2)
plt.pie(hours,labels=lang,autopct="%1.1f%%",explode=explode)

plt.subplot(2,2,3)
plt.scatter(hour,marks)

plt.show()
```
