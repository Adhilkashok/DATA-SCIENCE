```python
df.iloc[0,0]
```

```python
 #! filtering Data
#? marks>90 ?
# df[condition]
df[df["marks"]>90]  # df is used inside coz if not used it just shows true or false
```

```python
 #? marks>90 and age>20 ?
df[(df["marks"]>90) & (df["age"]>20)]
```

```python
 #? either marks>90 or age>20 ?
df[(df["marks"]>90) | (df["age"]>20)]
```

```python
 #? Exclude marks>90 ?
df[~(df["marks"]>90)]
```

```python
 #! Adding New Column
df["Grade"]=["B","A","A","","A"]
df
```

```python
 #! Creating new column from an existing column
df["new marks"]=df["marks"]*2
df
```

```python
 #! Modifying A column
df["marks"]=df["marks"]+2
df
```

```python
df["loc"]=df["loc"].str.upper()
df
```

```python
 #! Deleting columns
#! drop() 
df.drop("new marks",axis=1)   # deletes temporarily
# df=df.drop("new marks",axis=1)
```

```python
 #! inplace: retrieve dropped column
df.drop("new marks",axis=1,inplace=True)
```

```python
df
```

```python
df.drop("S2",inplace=True)
```

```python
df
```

```python
 #! reset_index -> Used to reset the index of a DataFrame back to the default numbers (0, 1, 2, 3, …)
df=df.reset_index(drop=True)
```

```python
df
```

```python
 #! adding new rows
new={"name":"Anumol",
     "age":19,
     "marks":85,
     "loc":"calicut",
     "Grade":"B"
     }
new_df=pd.DataFrame(new,index=["S6"])
```

```python
 #! concat() -> Used to combine two or more Pandas DataFrames or Series.
#! ignore_index=True → resets the index numbers after combining
df=pd.concat([df,new_df],ignore_index=True)
```

```python
import pandas as pd
dt={"name":["rajesh","arun","manoj","adhil","rohini"],
    "age":[26,29,28,26,25],
    "department":["IT","Finance","HR","IT","HR"],
    "salary":[28000,35000,32000,30000,25000]}
df3=pd.DataFrame(dt)
df3
   
```

```python
 #! unique(): Used to find the unique value in a column
df3["department"].unique()
```

```python
 #! nunique(): Returns the no.of unique value
df3["department"].nunique()
```

```python
 #! value_counts() -> Used to count how many times each unique value occurs in a Series/column
df3["department"].value_counts()
```

```python
 #! sum()
df3["salary"].sum()
```

```python
 #! mean()
df3["salary"].mean()
```

```python
 #! max()
df3["salary"].max()

 #! min()
df3["salary"].min()
```

```python
 #! var()
df3["salary"].var()
```
