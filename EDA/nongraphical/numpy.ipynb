```python
 #! OUTLIERS -> An outlier is a data value that is very different from the other values in a dataset.
# ---------------
#* eg: consider the age of students
# age=[19,22,26,200,28,32,8]
#* Here, 200 and 8 looks like an outlier because most ages are between 19 and 32, while 200 is much higher and 8 is much lower.
#* while calculating means we also consider these outlier points


#! IQR METHOD(INTER QUARTILE RANGE) -> METHODS TO DETECT OUTLIERS
# ----------
#! IQR=Q3-Q1
#! UPPER BOUND=Q3 + 1.5 * IQR
#! LOWER BOUND=Q1 - 1.5 * IQR
```

```python
age=np.array([19,22,26,200,28,32,5])
Q1=np.percentile(age,25)
Q3=np.percentile(age,75)
IQR=Q3-Q1
```

```python
lower=Q1-1.5*IQR
lower
```

```python
upper=Q3+1.5*IQR
upper
```

```python
outliers=age[(age>upper)|(age<lower)]
outliers
```

```python
 #! REMOVE OUTLIERS
new_age=age[(age>=lower) & (age<=upper)]
new_age
```

```python
 #! REPLACING WITH MEDIAN,MEAN
#! CAPPING METHOD -> REPLACING EXTREME VALUES WITH BOUNDARY VALUES
#?syntax: np.clip(array, lower_limit, upper_limit)
age=np.clip(age,5,45)
age
```

```python
 #! COVARIANCE -> Covariance tells whether two variables increase/decrease together or move in opposite directions.
#! COVARIANCE=COV(X,Y)=∑(Xi−Xˉ)(Yi​−Yˉ)/n-1

#! CORRELATION=COV(X,Y)/STD(X)*STD(Y)
```

```python
x=np.array([10,20,30,40])
y=np.array([15,25,35,45])
np.cov(x,y)

#* np.cov() returns a covariance matrix:
#* [[cov(x,x), cov(x,y)],
#*  [cov(y,x), cov(y,y)]]
```

```python
 #! To return the covariance btw x and y put [0,1]
np.cov(x,y)[0,1]
```

```python
 #! CORRELATION -1 TO 1
#* when correlation>0 to 1 -> positive correlation
#* when correlation=0 -> no linear correlation
#* when correlation<0 to -1 -> negative orrelation
np.corrcoef(x,y)[0,1]
```
