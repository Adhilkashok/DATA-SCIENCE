## DATA CLEANING


```python
 #! Handling Null Values
# ---------
 #! isnull() or isna() : Returns true if null
df.isnull()
```

```python
 #! column wise null value count
df.isnull().sum()
```

```python
 #! column wise percentage of null value
df.isnull().mean()*100
```

```python
 #! total null value count
df.isnull().sum().sum()
```

```python
df["City"].isnull().sum()
```

```python
df[df["City"].isnull()]
```

```python
df[df["Salesperson"].isnull()]
```

```python
 #! finding the most frequently occurring city in the City column.
mode_city=df["City"].mode()[0]
mode_city
```

```python
 #! fillna() -> Is used to replace missing values (NaN) in a DataFrame or Series.
df["City"]=df["City"].fillna(mode_city)
```

```python
df.isna().sum()
```

```python
df["Salesperson"]=df["Salesperson"].fillna(df["Salesperson"].mode()[0])
#? It is a categorical column, not a numerical column.
#? [0] means take the first value.
```

```python
df.isna().sum()
```

```python
df["Discount_Percent"]=df["Discount_Percent"].fillna(df["Discount_Percent"].median())
#? Median,Because Discount_Percent is numerical data, and the median is safer when there may be outliers.
```

```python
df.isna().sum()
```

```python
 #! ffill() -> It is used to fill missing (NaN) values using the previous available value (forward fill)
df.head()
```

```python
df.isna().sum()
```

```python
 #! bfill() -> It fills missing (NaN) values using the next available value.
df["Salesperson"]=df["Salesperson"].bfill()
```

```python
df.isna().sum()
```

```python
 #! dropna() -> Deletes rows having null value
df.dropna()   #axis=0-default
```

```python
df.dropna(axis=1)  #column
```

```python
import pandas as pd
one = pd.DataFrame({'id':[101,102,103,104],'Name':['Ammu','Sneha', 'Rasina','sajad'],
                    'sub_id':['sub1','sub2','sub3','sub4'],'Mark':[23,25,24,27]},index=[1,2,3,4])
two = pd.DataFrame({'id':[101,102,103,104],'Name':['Ammu1','Sneha1' ,'Rasina1','sajad1'],
                    'sub_id':['sub1','sub2','sub3','sub4'],'Mark':[24,28,29,19]},index=[1,2,3,4])
three = pd.DataFrame({'Name':['Ammu2','Sneha2','Rasina2', 'sajad2'],
                      'sub_id':['sub1','sub2','sub3','sub4'],
                      'Mark':[18,12,15,19],'roll_no':[12,13,14,15]},index=[1,2,3,4])
```

```python
pd.concat([one,two])
```

```python
pd.concat([one,two,three])
```

```python
pd.concat([one,two,three],keys=("x","y","z"))
```

```python
 #! ignore_index=True → resets the index numbers after combining
pd.concat([two,three],ignore_index=True)
```

```python
pd.concat([one,three],axis=1)   # concatenating column wise
```

```python
pd.merge(one,two,on=["id","sub_id"])
#! Means: Match a row only when both id AND sub_id are the same in both DataFrames.
```

```python
pd.merge(one,two,on=["sub_id"])
```

```python
 #! how ->types of join:left,right,outer,inner
pd.merge(one,two,on="sub_id",how="left")
#! Means Combine one and two using sub_id as the matching column, while keeping ALL rows from one.
```
