```python
x=df.iloc[:,:34]
x
```

```python
y=df.Target
y
```

```python
y.unique()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
ytrain.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE()

xtrain1,ytrain1=sm.fit_resample(xtrain,ytrain)

ytrain1.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
xtrain1=scaler.fit_transform(xtrain1)
xtrain1
```

```python
xtest=scaler.transform(xtest)
xtest
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=7)
model=knn.fit(xtrain1,ytrain1)
model
```

```python
ypred=knn.predict(xtest)
ypred
```

```python
ytest.values
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(ytest,ypred)
acc
```
