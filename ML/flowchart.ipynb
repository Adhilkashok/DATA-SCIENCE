### IMPORTANT

**SMOTE should be applied only to the training data, NOT to the test data.**

After Train-Test Split:

**Training Data → Feature Scaling → SMOTE → Model Training**

**Testing Data → Feature Scaling → Prediction → Accuracy**

**DO NOT APPLY SMOTE TO TEST DATA**, because it can cause **data leakage** and give an unreliable accuracy.
