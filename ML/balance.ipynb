# BALANCE:

#### **Balance** → Having a **similar number of samples in each class** of the target variable.

**Example:**

- YES → 500 samples
- NO → 500 samples

This is called a **Balanced Dataset**.

```python
import pandas as pd
df=pd.read_csv("/content/Copy of diabetes.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
x=df.iloc[:,:8]
x
```

```python
y=df.Outcome
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
```

```python
ytrain.value_counts()
```

# IMBALANCED DATA

**Imbalanced Data** → When the classes in the target variable do **not have a similar number of samples**.

### Example

- **Majority Class → 0 → 347 samples**
- **Minority Class → 1 → 190 samples**

Here, Class 0 has more samples than Class 1, so the data is **imbalanced**.

# DATA BALANCING

Data Balancing → The process of making the classes more balanced.

### 2 MAIN METHODS

### 1) OVERSAMPLING
- Increases the number of samples in the **minority class**.
- **SMOTE** creates new synthetic samples based on existing minority-class samples.
- **SMOTE → Synthetic Minority Over-sampling Technique**

### 2) UNDERSAMPLING
- Decreases the number of samples in the **majority class**.
- Some samples from the majority class are removed to make the classes more balanced.

```python
# SMOTE: Instead of simply duplicating minority samples, SMOTE creates new synthetic minority samples based on existing minority samples.
# Package=> imblearn
from imblearn.over_sampling import SMOTE
sm=SMOTE()

xtrain1,ytrain1=sm.fit_resample(xtrain,ytrain)
```

```python
ytrain1.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
xtrain1=scaler.fit_transform(xtrain1)
xtrain1
```

```python
xtest=scaler.transform(xtest)
xtest
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=7)
knn.fit(xtrain1,ytrain1)
ypred=knn.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(ytest,ypred)
acc
```
