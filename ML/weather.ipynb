```python
import pandas as pd
```

```python
df=pd.read_csv("/content/weatherData (1).csv",
               header=None,
               names=["Outlook","Temperature","Humidity","Wind","Play"])
df
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
## Outlook
# Encode 'Outlook' Column
le_o=LabelEncoder()
le1=le_o.fit(df["Outlook"])
df["Outlook"]=le1.transform(df["Outlook"])
df
```

```python
# View Original Labels for Outlook
le1.inverse_transform([0,1,2])
```

```python
# Temperature
le_t=LabelEncoder()
le2=le_o.fit(df["Temperature"])
df["Temperature"]=le2.transform(df["Temperature"])
df
```

```python
le2.inverse_transform([0,1,2])
```

```python
# Humidity
le_h=LabelEncoder()
le3=le_o.fit(df["Humidity"])
df["Humidity"]=le3.transform(df["Humidity"])
df
```

```python
le3.inverse_transform([0,1])
```

```python
# Wind
le_w=LabelEncoder()
le4=le_o.fit(df["Wind"])
df["Wind"]=le4.transform(df["Wind"])
df
```

```python
le4.inverse_transform([0,1])
```

```python
x=df.iloc[:,:4]
x
```

```python
y=df.Play
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
```

```python
from sklearn.naive_bayes import MultinomialNB
mnb=MultinomialNB()
mnb.fit(xtrain,ytrain)
```

```python
ypred=mnb.predict(xtest)
ypred
```

```python
y_new=mnb.predict([[
    le1.transform(["sunny"]).item(0),
    le2.transform(["cool"]).item(0),
    le3.transform(["high"]).item(0),
    le4.transform(["True"]).item(0)]])
y_new
```
