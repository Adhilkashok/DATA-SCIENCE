# SUPERVISED LEARNING → CLASSIFICATION

## 1. KNN (K-Nearest Neighbors)

- **Definition:** KNN is a supervised learning algorithm that predicts the output of a new data point by looking at its nearest data points.

- **Type:** Lazy Learner and Non-Parametric Algorithm.

- **Uses:** Classification and Regression (mainly used for classification).

### Important Points

- **Lazy Learner:** It does not learn a model during training. It stores the training data and uses it when making predictions.

- **Non-Parametric:** It does not assume that the data follows a specific mathematical distribution.

- **Classification:** Predicts a category by selecting the majority class among the K nearest neighbors.

- **Regression:** Predicts a numerical value using the values of the K nearest neighbors.

### Example
To predict whether a new fruit is an apple or an orange, KNN looks at the nearest fruits and assigns the class that appears most frequently among them.

## EUCLIDEAN DISTANCE FORMULA

Used to calculate the distance between two data points.

**Formula:**

d = √[(x₂ − x₁)² + (y₂ − y₁)²]

- (x₁, y₁) → Coordinates of the first data point.
- (x₂, y₂) → Coordinates of the second data point.
- d → Distance between the two data points.

## STEPS IN THE KNN ALGORITHM

**1) Calculate Distance**
- Calculate the distance between the new data point and all existing data points.

**2) Choose K**
- K represents the number of nearest neighbors to consider.
- K = 5 is a commonly used starting value, but the best value depends on the dataset.
- An odd value of K is often preferred in binary classification to reduce the chance of a tie.

**3) Find the K Nearest Neighbors**
- Sort the distances and select the K data points with the smallest distances.

**4) Count the Classes**
- Count how many neighbors belong to each class.

**Example:**

Suppose K = 3.

- YES = 1
- NO = 2

**5) Predict the Output**
- The majority class is NO.
- Therefore, the predicted output is **X′ = NO**.

```python
 #? QUESTION:
# POINTS    X1     Y1    TARGET
# P1        7      7      BAD
# P2        7      4      BAD
# P3        3      4      GOOD
# P4        1      4      GOOD
# P5:       X=3, Y=7 , TARGET=?

 #? ANSWER:
# d(P5 TO P1)=((3-7)^2 + (7-7)^2)^1/2 = 4
# d(P5 TO P2)=5
# d(P5 TO P3)=3
# d(P5 TO P4)=3.605
# HERE CLOSEST DATAS ARE P3=GOOD,P4=GOOD,P1=BAD
# GOOD IS GREATER THAN BAD
# THEREFORE, P5=GOOD
```

## TRAINING DATA AND TESTING DATA

- **Training Data (70%–80%):** The portion of the dataset used to train the machine learning model and help it learn patterns.

- **Testing Data (20%–30%):** The remaining portion used to evaluate how well the trained model performs on unseen data.

**Example:** If a dataset contains 1,000 records:
- Training Data (80%) → 800 records
- Testing Data (20%) → 200 records

**Note:** A common split is 80:20 or 70:30. Validation data is a separate portion used to tune the model's settings. Testing and validation data are not the same.

| **Input → X** | **Output → Y** |
|---|---|
| **X_train** | **Y_train** → 80% Training Data |
| **X_test** | **Y_test** → 20% Testing Data |

**X = Input Features (Independent Variables)**  
**Y = Output / Target (Dependent Variable)**
- **Training Data:** X_train + Y_train → 80%
- **Testing Data:** X_test + Y_test → 20%

### IMPORTANT TERMS:

- **X_train:** Input data used to train the model.
- **Y_train:** Actual output values used to train the model.
- **X_test:** Input data given to the trained model to make predictions.
- **Y_test:** Actual output values corresponding to X_test.
- **Y_pred:** Output values predicted by the trained model for X_test.

### MODEL EVALUATION

Compare **Y_test (actual output)** with **Y_pred (predicted output)** to evaluate the model's performance.

```python
#! INPUT and Output seperation
x=df.iloc[:,:8]
x
# x.ndim
# The dimension of input will always be 2Dimension

y=df.Outcome
y
# y.ndim
# The dimension of output will always be 1Dimension
```

```python
# Train and test split
# Package -> scikit learn/ sklearn
# ---------------
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
# test_size=0.30 means 30% of the data goes to the test set, and the remaining 70% goes to the training set.
# random_state controls how the data is randomly divided into training and testing sets -> 0,1,41,42,100
```

# FEATURE SCALING:

#### Feature Scaling → The process of bringing different numerical features to a similar scale before giving them to a Machine Learning model.

### 2 MAIN METHODS

#### 1) STANDARD SCALER (STANDARDIZATION)

- Data is centered around **0**.
- Changes the data so that **Mean = 0** and **Standard Deviation = 1**.
- Values can be **positive or negative**.

**Formula:**  
`X_new = (X_old − Mean) / Standard Deviation`

#### 2) MIN-MAX SCALER (NORMALIZATION)

- Changes the data into a fixed range, usually **0 to 1**.

**Formula:**  
`X_new = (X_old − X_min) / (X_max − X_min)`

### In Simple Words:

**StandardScaler → Data centered around 0**

**MinMaxScaler → Data changed to 0–1 range**

```python
# STANDARD SCALER METHOD
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
# fit() => Collect and calculate mean and std
scaler.fit(xtrain)
xtrain=scaler.transform(xtrain)
xtrain
```

```python
xtest=scaler.transform(xtest)
xtest
```

# KNN MODEL BUILDING

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier(n_neighbors=7)
model
# n_neighbors=7 means: For a new data point, KNN looks at its 7 nearest neighbors and uses them to decide the prediction.
```

```python
# Model Training
model.fit(xtrain,ytrain)
```

```python
# Testing
ypred=model.predict(xtest)
ypred
```

```python
ytest.values
```

## PERFORMANCE EVALUATION IN CLASSIFICATION: In classification, performance evaluation means checking how well our machine learning model predicts classes.
PE(ACTUAL O/P, PREDICTED O/P) --> PE(YTEST,YPRED)

METRICS==MODULE USED

1)CONFUSION MATRIX

2)ACCURACY SCORE

3)RECALL

4)PRECISION SCORE

5)F1 SCORE

### 1)CONFUSION MATRIX: A confusion matrix is a table used to evaluate a classification model by comparing: Actual values vs Predicted values.
M x N MATRIX

2 CLASS--> YES(1),NO(0)

If there are 2 classes, the confusion matrix is 2 × 2.

CM=2 x 2

2 x 2 -> 4 POINTS

1)TP (True Positive): Actual = 1, Predicted = 1

2)TN (True Negative): Actual = 0, Predicted = 0

3)FP (False Positive): Actual = 0, Predicted = 1

4)FN (False Negative): Actual = 1, Predicted = 0

```python
from sklearn.metrics import confusion_matrix
# syntax: cm=confusion_matrix(actual o/p,predicted o/p)
cm=confusion_matrix(ytest,ypred)
cm
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
```

## 2)ACCURACY SCORE: Accuracy tells us how many predictions were correct overall.

VALUE BTW 0 AND 1

ACC=(TP+TN)/(TP+TN+FP+FN)

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(ytest,ypred)
acc
```

## 3)RECALL: Recall tells us Out of all the actual positive cases, how many did the model correctly identify

. INVERSELY PROPORTIONAL TO FN

. R LOW --> FN HIGH

. R HIGH --> FN LOW

RECALL(1)=TP/(TP+FN)

RECALL(0)=TN/(TN+FP)

```python
from sklearn.metrics import  recall_score
recall=recall_score(ytest,ypred)
recall
```

## 4)PRECISION SCORE: Precision tells us:Out of all the cases the model predicted as positive, how many were actually positive

INVERSELY PROPORTIONAL TO FP

P(1)=TP/(TP=FP)

P(0)=TN/(TN+FN)

```python
from sklearn.metrics import precision_score
score=precision_score(ytest,ypred)
score
```

## 5)F1 SCORE: F1 Score tells us how well the model balances Precision and Recall.(COMBINATION OF RECALL AND PRECISION)

F1=2*([RECALL*PRECISION] / [RECALL+PRECISION])

```python
from sklearn.metrics import f1_score
score1=f1_score(ytest,ypred)
score1
```
