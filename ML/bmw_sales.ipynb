```python
x=df.iloc[:,:10]
x
```

```python
y=df.Sales_Classification
y
```

```python
x = pd.get_dummies(x, drop_first=True, dtype=int)
x
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

y = le.fit_transform(y)
y
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.3, random_state=41)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(xtrain)
xtrain = scaler.transform(xtrain)
xtrain
```

```python
xtest = scaler.transform(xtest)
xtest
```

```python
pd.Series(ytrain).value_counts()
```

```python
from imblearn.over_sampling import SMOTE
sm = SMOTE()
xtrain1, ytrain1 = sm.fit_resample(xtrain, ytrain)
```

```python
pd.Series(ytrain1).value_counts()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=7)
model=knn.fit(xtrain1,ytrain1)
model
```

```python
ypred=knn.predict(xtest)
ypred
```
