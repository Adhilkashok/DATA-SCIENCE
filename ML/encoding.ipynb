# ENCODING:

#### **Encoding** → The process of converting **categorical (text) data into numerical data** so that a Machine Learning model can understand and use it.

### 3 MAIN TYPES

#### 1) LABEL ENCODING
- Assigns a **number to each category**.
- Example: `Male → 0`, `Female → 1`

#### 2) ONE-HOT ENCODING
- Converts each category into a **separate column**.
- Uses **0 and 1** to represent the categories.

#### 3) GET DUMMIES
- `get_dummies()` is a **Pandas function** used to perform **One-Hot Encoding**.
- It automatically creates separate columns for each category.

---

### TYPES OF CATEGORICAL DATA

**Nominal Data** → Categories have **no natural order**.  
Example: `Red, Blue, Green`

**Ordinal Data** → Categories have a **meaningful order**.  
Example: `Low < Medium < High`
