## 2) NAIVE BAYES ALGORITHM:

#### **Naive Bayes** → A **supervised machine learning algorithm** mainly used for **classification**.

### KEY POINTS

- **Supervised Learning Algorithm**
- **Mainly used for Classification**
- Commonly used for **Text Classification**
- Based on **Probability** and **Bayes' Theorem**

### PROBABILITY

**Probability** → The chance of an event happening.

- Event → A
- Probability of A → **P(A)**

### CONDITIONAL PROBABILITY

**Conditional Probability** → The probability of an event happening **when another condition is already known**.

- Event → A
- Condition → B
- Conditional Probability → **P(A|B)**

### BAYES' THEOREM

Bayes' Theorem is used to calculate the probability of an event based on given information.

**Formula:**

`P(A|B) = [P(B|A) × P(A)] / P(B)`

**Where:**
- **P(A|B)** → Probability of A when B is known
- **P(B|A)** → Probability of B when A is known
- **P(A)** → Probability of A
- **P(B)** → Probability of B

```python
import pandas as pd
```

```python

df=pd.read_csv("/content/Copy of diabetes.csv")
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
x=df.iloc[:,:8]
x
```

```python
y=df.Outcome
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
```

## MODEL BUILDING:

### NAIVE BAYES MODELS:

**3 main types**:

### 1) GAUSSIAN NAIVE BAYES
- Used for **continuous numerical data**.
- Example: **Age, Height, Weight, Temperature**
- Think of it as **measurements**.

### 2) MULTINOMIAL NAIVE BAYES
- Used for **count or frequency data**.
- Commonly used for **text classification**.
- Example: **Number of times a word appears in a document**.

### 3) BERNOULLI NAIVE BAYES
- Used for **binary data (0 and 1)**.
- Represents whether something is **present/absent or yes/no**.
- Example: **Word present = 1, Word absent = 0**.

```python
from sklearn.naive_bayes import GaussianNB
model=GaussianNB()
```

```python
# MODEL TRAINING
model.fit(xtrain,ytrain)
```

```python
# VALIDATION
ypred=model.predict(xtest)
ypred
```

```python
# PE
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(ytest,ypred)
cm
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
cmd=ConfusionMatrixDisplay(cm)
cmd.plot()
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(ytest,ypred)
acc
```

```python
from sklearn.metrics import recall_score
recall=recall_score(ytest,ypred)
recall
```

```python
from sklearn.metrics import precision_score
score=precision_score(ytest,ypred)
score
```

```python
from sklearn.metrics import f1_score
score1=f1_score(ytest,ypred)
score1
```
