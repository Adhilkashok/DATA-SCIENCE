```python
num_cols = ["person_income", "person_emp_exp", "loan_amnt",
            "loan_int_rate", "loan_percent_income",
            "cb_person_cred_hist_length", "credit_score"]
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
cat_cols = ["person_education", "person_home_ownership","loan_intent", "previous_loan_defaults_on_file"]
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df=pd.get_dummies(df, drop_first=True,dtype=int)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

cols = ["person_age", "person_income", "person_emp_exp",
        "loan_amnt", "loan_int_rate", "loan_percent_income",
        "cb_person_cred_hist_length", "credit_score"]
for col in cols:
    plt.figure(figsize=(10, 8))
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df=df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
x=df.drop("loan_status",axis=1)
x
```

```python
y=df.loan_status
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
```

```python
ytrain.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE(random_state=41)
xtrain1,ytrain1=sm.fit_resample(xtrain,ytrain)
```

```python
ytrain1.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(xtrain)
xtrain1=scaler.transform(xtrain1)
xtest=scaler.transform(xtest)
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc=RandomForestClassifier(n_estimators=30,criterion="entropy",max_depth=3)
```

```python
rfc.fit(xtrain1,ytrain1)
```

```python
ypred=rfc.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest,ypred)
```

```python
# XGBOOST
from xgboost import XGBClassifier
xgb=XGBClassifier()
xgb.fit(xtrain1,ytrain1)
ypred=xgb.predict(xtest)
```

```python
accuracy_score(ytest,ypred)
```
