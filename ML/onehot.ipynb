```python
import pandas as pd
```

```python
df=pd.read_csv("/content/IRIS.csv")
df
```

```python
# One Hot encoder
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder
col_trns=make_column_transformer((OneHotEncoder(handle_unknown="ignore"),["species"]),remainder="passthrough")
                            # Basically says:"One-hot encode the species column and keep all the other columns unchanged."
```

```python
df1=col_trns.fit_transform(df)
# Means: "Learn the categories and transform the dataframe."
df1
```

```python
col_trns.get_feature_names_out()
# This tells you the names of the columns created by the transformer
# iris setosa | iris versicolar | iris virginica,.....
```
