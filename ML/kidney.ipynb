```python
for i in df.columns:
  print(i)
  print(df[i].unique())
  print("*"*50)
```

```python
df = df.drop("id", axis=1)
```

```python
df["pcv"] = df["pcv"].astype(float)
df["wc"] = df["wc"].astype(float)
df["rc"] = df["rc"].astype(float)
```

```python
df.isnull().sum()
```

```python
num_cols=["age", "bp", "sg", "al", "su", "bgr", "bu", "sc", "sod", "pot", "hemo", "pcv", "wc", "rc"]
for col in num_cols:
    df[col]=df[col].fillna(df[col].median())
```

```python
cat_cols=["rbc", "pc", "pcc", "ba", "htn", "dm","cad", "appet", "pe", "ane"]
for col in cat_cols:
    df[col]=df[col].fillna(df[col].mode()[0])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

for col in df.drop(columns="classification").columns:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
col=[ 'age', 'bp', 'sg', 'al', 'su', 'bgr',
       'bu', 'sc', 'sod', 'pot', 'hemo']
for i in col:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[i]=df[i].clip(lower,upper)
```

```python
for col in df.drop(columns="classification").columns:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
for j in df.columns:
  if df[j].dtype==object:
    df[j]=le.fit_transform(df[j])
```

```python
df.dtypes
```

```python
x=df.iloc[:,:-1]
x
```

```python
y=df.classification
y
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
```

```python
ytrain.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE()
xtrain1,ytrain1=sm.fit_resample(xtrain,ytrain)
```

```python
ytrain1.value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(xtrain)
xtrain1=scaler.transform(xtrain1)
xtest=scaler.transform(xtest)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(xtrain1, ytrain1)
ypred = knn.predict(xtest)
print(accuracy_score(ytest, ypred))

print("*"*50)

dt=DecisionTreeClassifier(criterion="entropy")
dt.fit(xtrain1, ytrain1)
ypred_dt=dt.predict(xtest)
print(accuracy_score(ytest, ypred_dt))

print("*"*50)

nb = GaussianNB()
nb.fit(xtrain1, ytrain1)
ypred_nb = nb.predict(xtest)
print(accuracy_score(ytest, ypred_nb))

print("*"*50)

rfc=RandomForestClassifier(n_estimators=30,criterion="entropy",max_depth=3)
rfc.fit(xtrain1,ytrain1)
ypred_rfc=rfc.predict(xtest)
print(accuracy_score(ytest,ypred_rfc))
```

```python
from xgboost import XGBClassifier
xgb=XGBClassifier()
xgb.fit(xtrain1,ytrain1)
ypred=xgb.predict(xtest)
accuracy_score(ytest,ypred)
```
