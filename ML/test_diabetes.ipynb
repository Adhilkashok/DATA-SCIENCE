```python
import seaborn as sns
import matplotlib.pyplot as plt

for col in df.drop(columns="Outcome").columns:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
col = ['Pregnancies', 'Glucose', 'BloodPressure',
       'SkinThickness', 'Insulin', 'BMI',
       'DiabetesPedigreeFunction', 'Age']
for i in col:
    Q1 = df[i].quantile(0.25)
    Q3 = df[i].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[i]=df[i].clip(lower,upper)
```

```python
for col in df.drop(columns="Outcome").columns:
  plt.figure(figsize=(10,8))
  sns.boxplot(x=df[col])
  plt.title(f"Boxplot of {col}")
  plt.show()
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=41)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(xtrain)
xtrain=scaler.transform(xtrain)
xtrain
```

```python
xtest=scaler.transform(xtest)
xtest
```

```python
ytrain.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
sm=SMOTE()
xtrain1,ytrain1=sm.fit_resample(xtrain,ytrain)
```

```python
ytrain1.value_counts()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(xtrain1, ytrain1)
ypred = knn.predict(xtest)
print(accuracy_score(ytest, ypred))

print("*"*50)

dt=DecisionTreeClassifier(criterion="entropy")
dt.fit(xtrain1, ytrain1)
ypred_dt=dt.predict(xtest)
print(accuracy_score(ytest, ypred_dt))

print("*"*50)

nb = GaussianNB()
nb.fit(xtrain1, ytrain1)
ypred_nb = nb.predict(xtest)
print(accuracy_score(ytest, ypred_nb))
```
